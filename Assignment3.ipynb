```python
#Write a python program that accepts the length of three sides of a triangle as inputs. 
#The program should indicate whether or not the triangle is a right-angled traingle using function.
```

```python
def right_angled_triangle(a,b,c):
    sides=sorted([a,b,c])
    if sides[2]**2==sides[0]**2+sides[1]**2:
        print("The triangle is a right-angled triangle.")
    else:
        print("The triangle is not a right-angled triangle.")
a=int(input("Enter first side of triangle:"))
b=int(input("Enter second side of triangle:"))
c=int(input("Enter third side of triangle:"))
right_angled_triangle(a,b,c)
```
