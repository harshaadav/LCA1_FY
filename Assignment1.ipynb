LISTS
Lists in Python are ordered, mutable collections that can hold heterogeneous items.

```python
a=[10,20,30]
a.append(40)#adds an element at the end of the list
print(a)
```

```python
a=[10,20,30]
a.insert(1,15)#Inserts an element at a specified position
print(a)
```

```python
a=[10,20,30,20]
a.remove(20)#Removes first occurence of a specified element
print(a)
```

```python
a=[10,20,30]
a.pop()#By default, removes the last element
print(a)
```

```python
a=[40,10,30,20]
a.sort()#Arranges the elements of a list in ascending order
print(a)
```

TUPLES
A tuple in Python is an immutable, ordered collection that can store elements of mixed data types.

```python
t=(1,2,2,3,2,4)
print(t.count(2))#Returns the number of times of a specified value appears in a tuple
```

```python
t=(10,20,30,40)
print(t.index(30))#Returns the index of the first occurence of an element
```

```python
t=(10,20,30,40)
print(len(t))#Returns the number of elements in a tuple
```

```python
t=(10,20,30,40,50)
print(t[1:4])#returns the elements which are present in the specific index bracket.
```

```python
t1=(10,20)
t2=(30,40)
t3=t1+t2#Two tuples can be joined by '+' sign
print(t3)
```

DICTIONARY
A dictionary in Python is a collection of key-value pairs, where each key is unique and maps to a value. 

```python
student={"name":"Rahul","age":18,"branch":"CSE"}
print(student.keys())#Returns all keys in the dictionary
```

```python
student={"name":"Rahul","age":18,"branch":"CSE"}
print(student.values())#Returns all the values
```

```python
student={"name":"Rahul","age":18}
print(student.items())#Returns all key-value pairs
```

```python
student={"name":"Rahul","age":18}
print(student.get("name"))#Returns the value associated with a specified key
```

```python
student={"name":"Rahul","age":18}
student.update({"age":19})#Modifies an existing value or adds a new key-value pair.
print(student)
```
