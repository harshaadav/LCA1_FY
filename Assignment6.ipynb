```python

import pandas as pd

# Read an existing CSV file
df = pd.read_csv('data.csv')

# For an Excel file, use:
# df = pd.read_excel('data.xls')

print("Original Dataset:")
print(df)

# 1. Display basic information
print("\nFirst Five Rows:")
print(df.head())

print("\nDataset Information:")
df.info()

# 2. Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Remove rows where all values are missing
df = df.dropna(how='all')

# Fill missing numerical values with their mean
for col in df.select_dtypes(include='number').columns:
    df[col] = df[col].fillna(df[col].mean())

# Fill missing text values with their mode
for col in df.select_dtypes(include=['str', 'object']).columns:
    if df[col].isnull().any() and not df[col].mode().empty:
        df[col] = df[col].fillna(df[col].mode()[0])

print("\nDataset After Handling Missing Values:")
print(df)

# 3. Identify duplicate records
print("\nDuplicate Records:")
print(df[df.duplicated()])

# Remove duplicate records
df = df.drop_duplicates()

print("\nDataset After Removing Duplicates:")
print(df)

# 4. Remove an irrelevant column
# Change 'Remarks' to an irrelevant column in your file
if 'Remarks' in df.columns:
    df = df.drop(columns=['Remarks'])

print("\nDataset After Removing Irrelevant Column:")
print(df)

# 5. Select rows and columns using loc[]
print("\nSelected Data:")
print(df.loc[df.index[:3]])

# Filter rows where a numerical column is greater than its mean
numeric_cols = df.select_dtypes(include='number').columns

if len(numeric_cols) > 0:
    col = numeric_cols[0]
    print(f"\nRows where {col} is above its mean:")
    print(df.loc[df[col] > df[col].mean()])

# 6. Save the cleaned dataset
df.to_csv('cleaned_data.csv', index=False)

print("\nData preprocessing completed successfully!")
```
